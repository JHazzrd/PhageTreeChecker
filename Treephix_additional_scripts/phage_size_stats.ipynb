```python
import pandas as pd
import argparse
from icecream import ic
```

```python
PATH = "Phage_data/"

df_meta = pd.read_excel(PATH+"VMR_MSL41.v1.20260320.xlsx", index_col = 23)

df_meta.head()
```

```python
ic(df_meta.at["PP467602","Genus"])
```

```python
df_phages = pd.read_csv(PATH+"7Apr2026_millardlab_website_table.tsv", sep = "\t")

df_phages.head()
```

```python
len('<a href="https://www.ncbi.nlm.nih.gov/nuccore/AY319521">')
```

```python
## Trim the Accession
def trim(x):
    x = x[56:-4]
    return x
df_phages["Accession"] = df_phages["Accession"].apply(trim)
```

```python
df_phages.head()
```

```python
df_phages.set_index("Accession", inplace=True)
```

```python
ic(df_phages.at["AY319521","Genome Length (KB)"])
```

```python
## Filter metadata for bacteriophages that are dsDNA

subset = df_meta[(df_meta["Genome"] == "dsDNA") & (df_meta["Host source"] == "bacteria")]

subset = subset["Genome"]
subset.head()
```

```python
subset.shape
```

```python
comb_df = pd.merge(subset, df_phages, how = "inner", on = "Accession")
```

```python
comb_df.head()
```

```python
comb_df.shape
```

```python
for col in comb_df.columns:
    print(col, comb_df[col].value_counts().count())
```

```python
genus_names = comb_df["Genus"].unique()
out_df = []
for g in genus_names:
    g_sub = comb_df[(comb_df["Genus"] == g)]
    med = g_sub["Genome Length (KB)"].median()
    m = g_sub["Genome Length (KB)"].min()
    mx = g_sub["Genome Length (KB)"].max()
    med_GC = g_sub["molGC (%)"].median()
    m_GC = g_sub["molGC (%)"].min()
    mx_GC = g_sub["molGC (%)"].max()
    s = g_sub.value_counts().count()
    d = {
        "Genus": g,
        "Median Genome Length (KB)": med,
        "Min Genome Length (KB)": m,
        "Max Genome Length (KB)": mx,
        "Median molGC (%)": med_GC,
        "Min molGC (%)": m_GC,
        "Max molGC (%)": mx_GC,
        "Sample Size": s
    }
    out_df.append(d)
out_df = pd.DataFrame(out_df)
```

```python
out_df.sort_values(["Sample Size"], ascending=[False]).head()
```

```python
out_df.to_csv("1612_phage_genome_stats.tsv",index = False, sep = "\t")
```
